
# implement a hidden markov model (HMM) to simulate phoneme transitions for the word ‘speech’


(a) Represent HMM Parameters

Initial Probabilities

Transition Matrix

Emission Matrix

```python
import random
import pandas as pd
```

```python
#hidden states (phonemes)
states = ["s", "p", "ie:", "tS"]

#observations
observations = ["Energy", "Pitch", "Duration"]

#initial probabilities (start at /s/)
pi = {
    "s": 1.0,
    "p": 0.0,
    "ie:": 0.0,
    "tS": 0.0
}

#transition probabilities for "speech"
#s -> p -> ie: -> tS
A = {
    "s":  {"s": 0.0, "p": 1.0, "ie:": 0.0, "tS": 0.0},
    "p":  {"s": 0.0, "p": 0.0, "ie:": 1.0, "tS": 0.0},
    "ie:": {"s": 0.0, "p": 0.0, "ie:": 0.0, "tS": 1.0},
    "tS": {"s": 0.0, "p": 0.0, "ie:": 0.0, "tS": 0.0},
}

#emission probabilities (given in question)
B = {
    "s":   {"Energy": 0.7, "Pitch": 0.2, "Duration": 0.1},
    "p":   {"Energy": 0.5, "Pitch": 0.3, "Duration": 0.2},
    "ie:": {"Energy": 0.3, "Pitch": 0.5, "Duration": 0.2},
    "tS":  {"Energy": 0.4, "Pitch": 0.4, "Duration": 0.2},
}
```

```python
#displaying matrices

def display_hmm(pi, A, B):
    print("\n--- Initial Probabilities ---")
    for state, prob in pi.items():
        print(f"{state} : {prob}")

    print("\n--- Transition Probability Matrix A ---")
    df_A = pd.DataFrame(A)
    print(df_A)

    print("\n--- Emission Probability Matrix B ---")
    df_B = pd.DataFrame(B)
    print(df_B)
```

```python
#sampling from a distribution


def sample_from_distribution(dist):
    items = list(dist.items())
    symbols = [i[0] for i in items]
    probs = [i[1] for i in items]
    return random.choices(symbols, weights=probs, k=1)[0]
```

```python
#to generate from one sequence
def generate_sequence(pi, A, B):
    phoneme_seq = []
    obs_seq = []

    #start state
    current = sample_from_distribution(pi)
    phoneme_seq.append(current)
    obs_seq.append(sample_from_distribution(B[current]))

    #continuing transitions until the last phoneme
    while True:
        if sum(A[current].values()) == 0:
            break   # end of sequence

        next_state = sample_from_distribution(A[current])
        phoneme_seq.append(next_state)
        obs_seq.append(sample_from_distribution(B[next_state]))

        current = next_state

    return phoneme_seq, obs_seq
```

```python
display_hmm(pi, A, B)

phonemes, observations_generated = generate_sequence(pi, A, B)

print("\nGenerated Phoneme Sequence:", phonemes)
print("Generated Observations:", observations_generated)
```
